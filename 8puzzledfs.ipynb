```python
print("Name: Lalit Kumar Sureliya")
print("USN: 1 W N 24 CS 144")

print("\n8 PUZZLE USING DFS")

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

visited = set()

def dfs(state):
    if state == goal:
        print("\nGoal reached!")
        print_board(state)
        return True

    visited.add(state)

    zero = state.index(0)
    row, col = zero // 3, zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c
            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            new_state = tuple(new_state)

            if new_state not in visited:
                print_board(new_state)

                if dfs(new_state):
                    return True

    return False

def print_board(state):
    print(state[0], state[1], state[2])
    print(state[3], state[4], state[5])
    print(state[6], state[7], state[8])
    print()

print("\nInitial State:")
print_board(start)

dfs(start)
```
