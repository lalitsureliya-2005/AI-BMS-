```python
import random
import math

N = 8

def heuristic(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j] or \
               abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def print_board(state):
    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
    print()


def simulated_annealing():
    while True:
        # Random initial state
        current = [random.randint(0, N - 1) for _ in range(N)]

        T = 1000.0
        cooling_rate = 0.99
        min_temperature = 0.01

        while T > min_temperature:
            h = heuristic(current)

            if h == 0:
                print("Solution Found!")
                print_board(current)
                return current

            # Generate a random neighboring state
            neighbor = current[:]
            col = random.randint(0, N - 1)
            row = random.randint(0, N - 1)
            neighbor[col] = row

            delta = heuristic(neighbor) - h

            # Accept better moves or sometimes worse moves
            if delta < 0 or random.random() < math.exp(-delta / T):
                current = neighbor

            T *= cooling_rate

        # If temperature becomes too low, restart

solution = simulated_annealing()

print("Queen positions:", solution)
print("Conflicts:", heuristic(solution))
print("Lalit Kumar Sureliya 1WN24CS144")
```
