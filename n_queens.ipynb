```python
import random

N = 4

def conflicts(board):
    count = 0
    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j] or \
               abs(board[i] - board[j]) == abs(i - j):
                count += 1
    return count


def print_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()
    print("Conflicts:", conflicts(board))
    print()


def hill_climbing():
    # Initial positions: row number of each column
    board = [0, 0, 0, 0]

    print("Initial Board:")
    print_board(board)

    while conflicts(board) > 0:
        current = conflicts(board)
        best_board = board[:]
        best_score = current
        best_move = None

        for col in range(N):
            old_row = board[col]

            for new_row in range(N):
                if new_row == old_row:
                    continue

                board[col] = new_row
                score = conflicts(board)

                if score < best_score:
                    best_score = score
                    best_board = board[:]
                    best_move = (col, old_row, new_row)

            board[col] = old_row

        if best_move is None:
            print("Local minimum reached. Restarting...")
            board = [random.randint(0, N - 1)
                     for _ in range(N)]
            print_board(board)
            continue

        col, old_row, new_row = best_move
        print(f"Move queen in column {col + 1}: "
              f"row {old_row + 1} -> row {new_row + 1}")

        board = best_board
        print_board(board)

    print("Solution Found!")


hill_climbing()
```
