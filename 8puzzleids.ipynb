```python
print("Name: Lalit Kumar Sureliya")
print("USN: 1WN24CS144")

print("\n8 PUZZLE USING DFS")

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

visited = set()

def show(s):
    print(s[0], s[1], s[2])
    print(s[3], s[4], s[5])
    print(s[6], s[7], s[8])
    print()

def dfs(state):
    if state == goal:
        print("Goal Reached!")
        show(state)
        return True

    visited.add(state)
    zero = state.index(0)
    row, col = zero // 3, zero % 3

    for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c
            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            new_state = tuple(new_state)

            if new_state not in visited:
                show(new_state)

                if dfs(new_state):
                    return True

    return False

print("\nInitial State:")
show(start)

dfs(start)
```
