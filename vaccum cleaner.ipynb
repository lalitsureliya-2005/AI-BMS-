```python
print("Name: Lalit Kumar Sureliya")
print("USN: 1 W N 24 CS 144")

print("\nVACUUM CLEANER")

rooms = {
    "A": "Dirty",
    "B": "Dirty"
}

location = input("Enter vacuum location (A/B): ").upper()

while True:
    print("\nRoom", location, "is", rooms[location])

    if rooms[location] == "Dirty":
        print("Cleaning room", location)
        rooms[location] = "Clean"
    else:
        print("Room", location, "is already clean")

    if all(room == "Clean" for room in rooms.values()):
        print("\nAll rooms are clean!")
        break

    location = "B" if location == "A" else "A"

print("Vacuum Cleaner stopped.")
```
