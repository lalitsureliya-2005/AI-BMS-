```python
def manhattan(puzzle, goal):
    distance = 0

    for i in range(3):
        for j in range(3):
            tile = puzzle[i][j]

            if tile != 0:
                for x in range(3):
                    for y in range(3):
                        if goal[x][y] == tile:
                            distance += abs(i - x) + abs(j - y)

    return distance


def print_matrix(puzzle):
    for row in puzzle:
        print(*row)
    print()


def get_moves(puzzle):
    moves = []

    for i in range(3):
        for j in range(3):
            if puzzle[i][j] == 0:
                x, y = i, j

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dx, dy in directions:
        a = x + dx
        b = y + dy

        if 0 <= a < 3 and 0 <= b < 3:
            new = [row[:] for row in puzzle]
            new[x][y], new[a][b] = new[a][b], new[x][y]
            moves.append(new)

    return moves


goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

puzzle = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

initial_distance = manhattan(puzzle, goal)

print("Initial State:")
print_matrix(puzzle)

g = 0

while puzzle != goal:

    possible = get_moves(puzzle)

    best = None
    best_f = 999

    for move in possible:
        h = manhattan(move, goal)
        f = g + 1 + h

        if f < best_f:
            best_f = f
            best = move

    puzzle = best
    g += 1

    print("Step", g)
    print_matrix(puzzle)

print("Goal State:")
print_matrix(puzzle)

print("Initial Manhattan Distance =", initial_distance)
print("Total Steps =", g)
```
