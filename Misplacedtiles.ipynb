```python
def misplaced(puzzle, goal):
    count = 0
    for i in range(3):
        for j in range(3):
            if puzzle[i][j] != 0 and puzzle[i][j] != goal[i][j]:
                count += 1
    return count


def print_matrix(puzzle):
    for row in puzzle:
        print(*row)
    print()


def get_moves(puzzle):
    moves = []

    for i in range(3):
        for j in range(3):
            if puzzle[i][j] == 0:
                x, y = i, j

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dx, dy in directions:
        a = x + dx
        b = y + dy

        if 0 <= a < 3 and 0 <= b < 3:
            new = [row[:] for row in puzzle]
            new[x][y], new[a][b] = new[a][b], new[x][y]
            moves.append(new)

    return moves


goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

puzzle = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

initial_misplaced = misplaced(puzzle, goal)

print("Initial State:")
print_matrix(puzzle)

while puzzle != goal:

    possible = get_moves(puzzle)

    best = min(possible, key=lambda x: misplaced(x, goal))

    print("Next Step:")
    print_matrix(best)

    puzzle = best

print("Goal State:")
print_matrix(puzzle)

print("Total Misplaced Tiles in Initial State =", initial_misplaced)
```
