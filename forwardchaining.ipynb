```python

def forward_chaining(facts, rules, goal):

    facts = set(facts)

    while True:
        new_fact = False

        for conditions, conclusion in rules:

            # Check whether all conditions are satisfied
            if all(condition in facts for condition in conditions):

                # Add conclusion if it is not already present
                if conclusion not in facts:

                    facts.add(conclusion)
                    print("New fact derived:", conclusion)

                    new_fact = True

                    # Check whether goal is reached
                    if conclusion == goal:
                        return True

        # Stop if no new facts are generated
        if not new_fact:
            break

    return goal in facts


# Initial facts
facts = [
    "Barks(Tommy)",
    "FourLegs(Tommy)",
    "SharpTeeth(Tommy)"
]

# Rules: (conditions, conclusion)
rules = [

    (["Barks(Tommy)", "FourLegs(Tommy)"],
     "Dog(Tommy)"),

    (["HasFeathers(Tommy)", "LaysEggs(Tommy)"],
     "Bird(Tommy)"),

    (["Dog(Tommy)"],
     "Mammal(Tommy)"),

    (["Mammal(Tommy)", "SharpTeeth(Tommy)"],
     "Carnivore(Tommy)"),

    (["Bird(Tommy)"],
     "CanFly(Tommy)")
]

# Goal
goal = "Carnivore(Tommy)"

# Execute Forward Chaining
if forward_chaining(facts, rules, goal):
    print("\nGoal achieved: by LALIT KUMAR SURELIYA(1WN24CS144)-->", goal)
else:
    print("\nGoal cannot be proved")
```
