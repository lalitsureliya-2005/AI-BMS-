```python
def unify(x, y):
    subs = {}

    for a, b in zip(x, y):
        if a == b:
            continue
        elif a.islower():
            subs[a] = b
        elif b.islower():
            subs[b] = a
        else:
            return "Unification Failed"

    return subs


# Example
x = ["P", "x", "a"]
y = ["P", "b", "a"]

result = unify(x, y)
print("Substitution:", result)
print("LALIT KUMAR SURELIYA (1WN24CS144)")
```
