```python

from itertools import product

# Truth-table enumeration algorithm
def TT_ENTAILS(KB, query, symbols):
    for values in product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))

        if KB(model) and not query(model):
            return False

    return True


# Knowledge Base:
# (P <-> Q) AND (Q OR R) AND (NOT P -> R)
def KB(model):
    P = model['P']
    Q = model['Q']
    R = model['R']

    sentence1 = (P == Q)             # P <-> Q
    sentence2 = Q or R                # Q OR R
    sentence3 = P or R                # NOT P -> R

    return sentence1 and sentence2 and sentence3


# Queries
def query_Q(model):
    return model['Q']


def query_P_implies_Q(model):
    return (not model['P']) or model['Q']


def query_R_implies_Q(model):
    return (not model['R']) or model['Q']


# Display truth table
symbols = ['P', 'Q', 'R']

print("TRUTH TABLE")
print("P\tQ\tR\tKB")

models = []

for values in product([True, False], repeat=len(symbols)):
    model = dict(zip(symbols, values))
    kb_value = KB(model)

    print(
        model['P'], "\t",
        model['Q'], "\t",
        model['R'], "\t",
        kb_value
    )

    if kb_value:
        models.append(model)

print("\nMODELS OF KB:")
for model in models:
    print(model)

# Check all required entailment cases
queries = [
    ("KB entails Q", query_Q),
    ("KB entails P -> Q", query_P_implies_Q),
    ("KB entails R -> Q", query_R_implies_Q)
]

print("\nENTAILMENT RESULTS:")

for description, query in queries:
    if TT_ENTAILS(KB, query, symbols):
        print(description + ": TRUE")
    else:
        print(description + ": FALSE")

print("\nLALIT KUMAR SURELIYA, 1WN24CS144")
```
